```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

data_jan = [
    (1, "Alice", 5000),
    (2, "Bob", 6000),
    (3, "Charlie", 5500)
]

schema = StructType([
    StructField("emp_id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("salary", IntegerType(), True)
])

df_jan = spark.createDataFrame(data_jan, schema)
display(df_jan)
```

```python
data_feb = [
    (4, "David", 6500),
    (5, "Eva", 7000)
]

df_feb = spark.createDataFrame(data_feb, schema)
display(df_feb)
```

```python
data_mar = [
    ("Frank", 7500, 6),
    ("Grace", 8000, 7)
]

schema_mar = StructType([
    StructField("name", StringType(), True),
    StructField("salary", IntegerType(), True),
    StructField("emp_id", IntegerType(), True)
])

df_mar = spark.createDataFrame(data_mar, schema_mar)
display(df_mar)
```

```python
data_apr = [
    (8, "Henry", 9000, 1000)
]

schema_apr = StructType([
    StructField("emp_id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("salary", IntegerType(), True),
    StructField("bonus", IntegerType(), True)
])

df_apr = spark.createDataFrame(data_apr, schema_apr)
display(df_apr)
```

```python
df_jan.union(df_feb).display()
```

```python
df_jan.unionByName(df_mar).display()
```

```python
df_jan.unionByName(df_apr, allowMissingColumns=True).display()
```
